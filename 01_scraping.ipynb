```python
import os
import requests
import time

save_folder = "../datasets/city_images_dataset1"
os.makedirs(save_folder, exist_ok=True)

downloaded = 0

for i in range(300):   # 请求300张，保证够用
    try:
        url = f"https://picsum.photos/400/300?random={i}"
        
        img_data = requests.get(url, timeout=10).content
        
        file_path = os.path.join(save_folder, f"city_{downloaded}.jpg")
        
        with open(file_path, "wb") as f:
            f.write(img_data)

        downloaded += 1

        if downloaded % 20 == 0:
            print("Downloaded:", downloaded)

        time.sleep(0.1)

    except Exception as e:
        print("Error:", e)

print("Final downloaded:", downloaded)
```

```python
import os
import requests
import time
from PIL import Image

save_folder = "../datasets/wikimedia_city_street_images"
os.makedirs(save_folder, exist_ok=True)

API_URL = "https://commons.wikimedia.org/w/api.php"

headers = {
    "User-Agent": "DfPI student project / city street image dataset"
}

search_terms = [
    "city street",
    "urban street",
    "street scene city",
    "pedestrian street",
    "city road",
    "urban public space",
    "cityscape street",
    "downtown street"
]

image_urls = []

for term in search_terms:
    params = {
        "action": "query",
        "format": "json",
        "generator": "search",
        "gsrsearch": term,
        "gsrnamespace": 6,
        "gsrlimit": 50,
        "prop": "imageinfo",
        "iiprop": "url",
        "iiurlwidth": 800
    }

    response = requests.get(API_URL, params=params, headers=headers)
    data = response.json()

    pages = data.get("query", {}).get("pages", {})

    for page_id, page in pages.items():
        if "imageinfo" in page:
            info = page["imageinfo"][0]
            if "thumburl" in info:
                image_urls.append(info["thumburl"])

print("Image URLs found:", len(image_urls))
```

```python
downloaded = 0

for url in image_urls:
    if downloaded >= 250:
        break

    try:
        response = requests.get(url, timeout=10, headers=headers)
        file_path = os.path.join(save_folder, f"street_{downloaded}.jpg")

        with open(file_path, "wb") as f:
            f.write(response.content)

        # 验证是不是真图片
        img = Image.open(file_path)
        img.verify()

        downloaded += 1

        if downloaded % 20 == 0:
            print("Downloaded:", downloaded)

        time.sleep(0.2)

    except Exception as e:
        if os.path.exists(file_path):
            os.remove(file_path)

print("Final downloaded:", downloaded)
```

```python
import os
import requests
import json
import time

save_folder = "../datasets/openlibrary_urban_texts"
os.makedirs(save_folder, exist_ok=True)

keywords = [
    "urban design", "architecture", "city", "street", "urban planning",
    "public space", "city planning", "building", "urban landscape",
    "urban morphology", "city streets", "housing", "modern architecture",
    "urban studies", "pedestrian", "road", "town planning", "city life"
]

texts = []

for keyword in keywords:
    url = "https://openlibrary.org/search.json"
    params = {
        "q": keyword,
        "limit": 100
    }

    response = requests.get(url, params=params, timeout=20)
    data = response.json()

    for book in data.get("docs", []):
        title = book.get("title", "")
        author = ", ".join(book.get("author_name", [])[:2])
        subjects = ", ".join(book.get("subject", [])[:8])

        text = f"Title: {title}. Author: {author}. Subjects: {subjects}."

        if title and text not in [t["text"] for t in texts]:
            texts.append({
                "keyword": keyword,
                "title": title,
                "text": text
            })

        if len(texts) >= 250:
            break

    print(keyword, "current total:", len(texts))

    if len(texts) >= 250:
        break

    time.sleep(0.2)

with open(f"{save_folder}/urban_books_texts.json", "w", encoding="utf-8") as f:
    json.dump(texts[:250], f, ensure_ascii=False, indent=2)

print("Open Library text items saved:", len(texts[:250]))
```
